This notebook contains the code to find the attractors (stable states) for the rules of the inferred network. The function `find_attrs` takes a rule as input, builds a succession diagram and returns a dataframe with the attractor seeds.

```python
import pickle
import biobalm as balm
import pandas as pd
```

```python
def find_attrs(rule):
    sd = balm.SuccessionDiagram.from_rules(rule)
    sd.build()
    data = sd.expanded_attractor_seeds()
# convert dict into dataframe
    flat = {k: v[0] for k, v in data.items()}
    df_ss = pd.DataFrame.from_dict(flat, orient="index")
    return df_ss
```
